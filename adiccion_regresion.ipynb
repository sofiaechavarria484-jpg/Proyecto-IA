### Análisis de Codificación para este Dataset

Para este dataset específico de estudiantes, el **mejor método es una combinación**, pero si debemos elegir uno dominante, el **Target Encoding** o **Label Encoding** (solo para stress) son superiores por las siguientes razones:

1.  **Stress_level (Ordinal):** El mejor método es **Label Encoding** (o mapeo manual), ya que el nivel de estrés tiene un orden lógico (Low < Medium < High) que este método preserva.
2.  **Gender (Binario):** Al tener solo dos categorías, Label Encoding o One-Hot dan resultados similares sin añadir complejidad.
3.  **Gaming_genre (Nominal):** Aquí el **Target Encoding** es muy potente porque captura la relación directa entre el género de juego y el rendimiento académico (`grades`) sin aumentar la dimensionalidad del dataset (como lo haría One-Hot con muchas columnas nuevas).

**Conclusión:** Recomiendo usar **Label Encoding** para `stress_level` y **One-Hot** para los demás si buscas interpretabilidad, o **Target Encoding** si el objetivo principal es maximizar la precisión del modelo de regresión.

```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import LabelEncoder

# Crear copias del dataframe original para cada método
df_label = df.copy()
df_oh = df.copy()
df_target = df.copy()

cat_cols = ['gender', 'gaming_genre', 'stress_level']

# 1. Label Encoding
# Útil para variables con orden (como stress_level) o binarias
le = LabelEncoder()
for col in cat_cols:
    df_label[col] = le.fit_transform(df_label[col])

# 2. One-Hot Encoding
# Crea columnas binarias para cada categoría. Evita sesgos de orden.
df_oh = pd.get_dummies(df_oh, columns=cat_cols)

# 3. Target Encoding (basado en la media de 'grades')
# Reemplaza la categoría por el promedio de la variable objetivo
for col in cat_cols:
    target_means = df.groupby(col)['grades'].mean()
    df_target[col] = df_target[col].map(target_means)

print("Codificaciones completadas:")
print(f"- df_label: {df_label.shape}")
print(f"- df_oh: {df_oh.shape}")
print(f"- df_target: {df_target.shape}")

# --- EXPLICACIÓN DEL MEJOR MÉTODO ---
"""
¿Cuál es el mejor método para este dataset?

Para este dataset de rendimiento académico, la mejor estrategia es un ENFOQUE HÍBRIDO:

1. Label Encoding para 'stress_level': Es el mejor porque 'stress_level' es una variable ORDINAL
   (Low < Medium < High). El Label Encoding permite que el modelo entienda esta jerarquía.

2. Target Encoding para 'gaming_genre': Al ser una variable nominal con varias categorías,
   el Target Encoding captura directamente cómo cada género afecta al promedio de 'grades'
   sin aumentar la cantidad de columnas (evitando la 'maldición de la dimensionalidad'
   que causaría el One-Hot encoding).

3. One-Hot o Label para 'gender': Al ser binaria (Male/Female), cualquiera funciona bien.

Conclusión: Si tienes que elegir un solo dataframe para un modelo de regresión,
usa 'df_target' para capturar relaciones complejas, o 'df_label' si quieres simplicidad.
"""
```

```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, PolynomialFeatures
from sklearn.linear_model import LinearRegression, Ridge, Lasso, ElasticNet
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import RandomForestRegressor, GradientBoostingRegressor
from sklearn.svm import SVR
from sklearn.neighbors import KNeighborsRegressor
from sklearn.metrics import r2_score, mean_squared_error

# --- RECONSTRUCCIÓN DE DEPENDENCIAS ---
# Cargamos y preparamos df_target para evitar el NameError
file_id = '1z10sx-JCQwR09p_HCgSJdX_JdFEIug_R'
url = f'https://drive.google.com/uc?export=download&id={file_id}'
df = pd.read_csv(url)

df_target = df.copy()
cat_cols = ['gender', 'gaming_genre', 'stress_level']
for col in cat_cols:
    target_means = df.groupby(col)['grades'].mean()
    df_target[col] = df_target[col].map(target_means)

# --- INICIO DE EXPERIMENTOS ---
X = df_target.drop(columns=['student_id', 'grades'])
y = df_target['grades']

splits = [0.30, 0.25, 0.20]
results = []

def get_models():
    return {
        'Linear Regression': LinearRegression(),
        'Ridge': Ridge(),
        'Lasso': Lasso(),
        'Elastic Net': ElasticNet(),
        'Polynomial (Deg 2)': LinearRegression(),
        'Decision Tree': DecisionTreeRegressor(random_state=42),
        'Random Forest': RandomForestRegressor(n_estimators=100, random_state=42),
        'Gradient Boosting': GradientBoostingRegressor(random_state=42),
        'SVR': SVR(),
        'KNN': KNeighborsRegressor()
    }

print("Iniciando experimentos...")

for test_size in splits:
    train_size = int((1 - test_size) * 100)
    print(f"Evaluando split {train_size}-{int(test_size*100)}...")

    X_train_raw, X_test_raw, y_train, y_test = train_test_split(X, y, test_size=test_size, random_state=42)

    scaler = StandardScaler()
    X_train = scaler.fit_transform(X_train_raw)
    X_test = scaler.transform(X_test_raw)

    models = get_models()

    for name, model in models.items():
        if name == 'Polynomial (Deg 2)':
            poly = PolynomialFeatures(degree=2)
            X_train_p = poly.fit_transform(X_train)
            X_test_p = poly.transform(X_test)
            model.fit(X_train_p, y_train)
            preds = model.predict(X_test_p)
        else:
            model.fit(X_train, y_train)
            preds = model.predict(X_test)

        r2 = r2_score(y_test, preds)
        rmse = np.sqrt(mean_squared_error(y_test, preds))

        results.append({
            'Split': f"{train_size}/{int(test_size*100)}",
            'Algoritmo': name,
            'R2 Score': round(r2, 4),
            'RMSE': round(rmse, 4)
        })

df_results = pd.DataFrame(results)
print("\n--- RESUMEN DE EXPERIMENTOS ---")
display(df_results.sort_values(by=['Split', 'R2 Score'], ascending=[True, False]))
```

```python
import pandas as pd
import numpy as np
from sklearn.ensemble import GradientBoostingRegressor
from sklearn.preprocessing import StandardScaler

# 1. Preparar el modelo ganador con los datos de df_target
X = df_target.drop(columns=['student_id', 'grades'])
y = df_target['grades']

# Entrenamos el escalador y el modelo final
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

best_model = GradientBoostingRegressor(random_state=42)
best_model.fit(X_scaled, y)

# 2. Diccionarios para mapear categorías (Target Encoding)
target_maps = {}
for col in ['gender', 'gaming_genre', 'stress_level']:
    target_maps[col] = df.groupby(col)['grades'].mean().to_dict()

print("--- SIMULADOR DE PREDICCIÓN DE GRADES ---")
try:
    # Entradas del usuario
    age = float(input("Edad (ej. 20): "))
    gender = input("Género (Male/Female/Other): ")
    gaming_hours = float(input("Horas de juego diarias: "))
    study_hours = float(input("Horas de estudio diarias: "))
    sleep_hours = float(input("Horas de sueño diarias: "))
    attendance = float(input("Asistencia % (0-100): "))
    genre = input("Género de juego (FPS/RPG/Casual/Sports/Action/Other): ")
    social = float(input("Actividad social (1-5): "))
    device = float(input("Uso de dispositivos (horas): "))
    reaction = float(input("Tiempo de reacción (ms): "))
    addiction = float(input("Score de adicción (0-25): "))
    stress = input("Nivel de estrés (Low/Medium/High): ")

    # Transformación de entrada
    user_input = pd.DataFrame([[
        age,
        target_maps['gender'].get(gender, df['grades'].mean()),
        gaming_hours, study_hours, sleep_hours, attendance,
        target_maps['gaming_genre'].get(genre, df['grades'].mean()),
        social, device, reaction, addiction,
        target_maps['stress_level'].get(stress, df['grades'].mean())
    ]], columns=X.columns)

    # Escalar y predecir
    user_scaled = scaler.transform(user_input)
    prediction = best_model.predict(user_scaled)[0]

    print(f"\n--- PREDICCIÓN FINAL ---")
    print(f"La calificación estimada para el estudiante es: {prediction:.2f}")

except Exception as e:
    print(f"Error en el ingreso de datos: {e}")
```
